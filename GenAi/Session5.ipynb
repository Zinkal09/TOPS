Ques 2.
Create a simple Q&A chatbot for Zomato-style restaurant info using any language or platform you know (Python, JavaScript, etc.): given a restaurant name and cuisine, return a canned response like 'Yes, Pizza Palace serves Italian cuisine.'

```python
# Zomato-style Restaurant Chatbot

# Restaurant database
restaurants = {
    "Pizza Palace": "Italian",
    "Spice Villa": "Indian",
    "Dragon House": "Chinese",
    "Taco Town": "Mexican",
    "Burger Hub": "American"
}

# User input
restaurant_name = input("Enter restaurant name: ")
cuisine = input("Enter cuisine: ")

# Check if restaurant exists
if restaurant_name in restaurants:
    if restaurants[restaurant_name].lower() == cuisine.lower():
        print(f"Yes, {restaurant_name} serves {cuisine} cuisine.")
    else:
        print(f"No, {restaurant_name} serves {restaurants[restaurant_name]} cuisine.")
else:
    print(f"Sorry, I don't have information about {restaurant_name}.")
```

Ques 4: Use GitHub Copilot or ChatGPT to help you write a function that takes a list of song names and returns only those that start with the letter 'S', like a Spotify playlist filter.<br><br><em><strong>Constraint:</strong> You must show the prompt you used and the code Copilot or ChatGPT generated.</em>


```python
def filter_songs_starting_with_s(song_list):
    """Return songs that start with the letter 'S'."""
    return [song for song in song_list if song.upper().startswith('S')]

# Example playlist
songs = [
    "Shape of You",
    "Sunflower",
    "Believer",
    "Senorita",
    "Perfect",
    "Someone Like You",
    "Thunder"
]

# Filter songs
filtered_songs = filter_songs_starting_with_s(songs)

# Display result
print("Songs starting with 'S':")
print(filtered_songs)
```
