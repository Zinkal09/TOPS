```python
#Q.2 Get the common items between a and b
a = np.array([1,2,3,2,3,4,3,4,5,6,5,8,2,1,2,3])
b = np.array([7,2,10,2,7,4,9,4,9,8,23,4,56,1,2])

common = np.intersect1d(a, b)
print("Common items:", common)
```

```python
# Q.3 Get all items between 5 and 10 from a.
a = np.array([2, 6, 1, 9, 10, 3, 27])

result = np.where((a >= 5) & (a <= 10))
print("Values:", a[result])
```

```python
# Q1. Compute the minimum, 25th percentile, median, 75th, and maximum of ser. 
s1 = pd.Series([12, 7, 3, 18, 9,24,2,5,11,23, 21, 5, 14, 8, 11])

result = pd.Series({
    "min": s1.min(),
    "25%": s1.quantile(0.25),
    "median": s1.median(),
    "75%": s1.quantile(0.75),
    "max": s1.max()
})
print(result)
```

```python
#Q9. Create a line plot using the x and y values provided below. 
#Label the y-axis as “Y” and label the x-axis as “X”. x = [3, 4, 5, 6] y = [1.5, 2, 2.5, 3] 
x = [3, 4, 5, 6]
y = [1.5, 2, 2.5, 3]

plt.plot(x, y, marker="o")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Line Plot")
plt.show()
```

```python
#Q10. Create an array of numbers between 0 and 6 with increments of 0.3 and name its “x”.
#Then on the same plot, plot x, x², x³, and x⁴. For consistency, use the following style lines respectively, “ro”, “bs”, “g”, and “:”
# Lastly, make sure that the x-axis covers 0 to 6, while the y-axis spans from 0 to 125. 
#Do not worry if you are not familiar with the style lines — you will recognize them as soon as you see the plot. 

x = np.arange(0, 6.1, 0.3)

plt.plot(x, x**2, "ro", label="x^2")   # red circles
plt.plot(x, x**3, "bs", label="x^3")   # blue squares
plt.plot(x, x**4, "g", label="x^4")    # green line
plt.plot(x, x, ":", label="x")         # dotted line

plt.xlim(0, 6)
plt.ylim(0, 125)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Polynomial Functions")
plt.legend()
plt.show()
```

```python
#11. Heights and initials of a group of individuals are provided below.
#Create a bar plot titled “Height Comparison” to compare the heights among this group.
#height = [179, 155, 191, 152, 188, 177] names = ['QA', 'WB', 'EC', 'RD', 'TE', 'YF']
height = [179, 155, 191, 152, 188, 177]
names = ['QA', 'WB', 'EC', 'RD', 'TE', 'YF']

plt.bar(names, height, color="skyblue")
plt.title("Height Comparison")
plt.xlabel("Names")
plt.ylabel("Height (cm)")
plt.show()
```

```python
#12. Plot a histogram of x, where x consists of 100,000 randomly-selected points with a normal distribution 
#(hint: you can use numpy.random.randn() to generate the random points). 
#The histogram should have 10 bins. Look at how the histogram changes when we try 20 and 50 bins. 

x = np.random.randn(100000)  # 100k points

# Histogram with 10 bins
plt.hist(x, bins=10, color="skyblue", edgecolor="black")
plt.title("Histogram with 10 bins")
plt.show()

# Histogram with 20 bins
plt.hist(x, bins=20, color="lightgreen", edgecolor="black")
plt.title("Histogram with 20 bins")
plt.show()

# Histogram with 50 bins
plt.hist(x, bins=50, color="orange", edgecolor="black")
plt.title("Histogram with 50 bins")
plt.show()
```
