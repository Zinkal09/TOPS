```python
import matplotlib.pyplot as plt
```

```python
image= cv2.imread(r"apple.jpg")
# Convert BGR to RGB for display
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Resize dimensions
new_size = (700, 700)

# Resize using INTER_LINEAR
linear_resize = cv2.resize(
    image,
    new_size,
    interpolation=cv2.INTER_LINEAR
)
cubic_resize = cv2.resize(
    image,
    new_size,
    interpolation=cv2.INTER_CUBIC
)
```

```python
difference = cv2.absdiff(linear_resize, cubic_resize)
```

```python
# Convert images for plotting
linear_rgb = cv2.cvtColor(linear_resize, cv2.COLOR_BGR2RGB)
cubic_rgb = cv2.cvtColor(cubic_resize, cv2.COLOR_BGR2RGB)
diff_rgb = cv2.cvtColor(difference, cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=(15,5))
```

```python
plt.subplot(1,3,1)
plt.imshow(linear_rgb)
plt.title("INTER_LINEAR")
plt.axis("off")
```

```python
plt.subplot(1,3,2)
plt.imshow(cubic_rgb)
plt.title("INTER_CUBIC")
plt.axis("off")
```

```python
plt.subplot(1,3,3)
plt.imshow(diff_rgb)
plt.title("Pixel-Level Difference")
plt.axis("off")

plt.show()
```

```python
image= cv2.imread(r"apple.jpg")
# Convert BGR to RGB for display
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Resize dimensions
new_size = (1200, 1200)

# Resize using INTER_LINEAR
linear_resize = cv2.resize(
    image,
    new_size,
    interpolation=cv2.INTER_LINEAR
)
cubic_resize = cv2.resize(
    image,
    new_size,
    interpolation=cv2.INTER_CUBIC
)
```

```python
img = cv2.imread(r"apple.jpg")
h,w = img.shape[:2]

cv2.imshow("original",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python

center = (w // 2, h // 2)   # Find exact center of image
angle = 45                   # Degrees to rotate (positive = counter-clockwise)
scale = 1.0 

M = cv2.getRotationMatrix2D(center, angle, scale)
rotated = cv2.warpAffine(img, M, (w, h))
cv2.imshow("Rotated", rotated)
cv2.waitKey(0)

cv2.destroyAllWindows()
```

```python
center = (w // 2, h // 2)   # Find exact center of image
angle = 90                   # Degrees to rotate (positive = counter-clockwise)
scale = 1.0 

M = cv2.getRotationMatrix2D(center, angle, scale)
rotated = cv2.warpAffine(img, M, (w, h))
cv2.imshow("Rotated", rotated)
cv2.waitKey(0)

cv2.destroyAllWindows()
```

```python
flipped_h = cv2.flip(img, 1)   # flipCode=1  - horizontal (mirror left-right)
flipped_v = cv2.flip(img, 0)
```

```python
import cv2

# Load image
img = cv2.imread(r"C:\Users\Zinkal Patel\Desktop\Images\LinuxLogo.jpg")

# ---------------------------------------------------
# ROI CROP
# ---------------------------------------------------

# ROI coordinates
x1, y1 = 100, 100
x2, y2 = 300, 300

# Extract ROI
roi = img[y1:y2, x1:x2]

# ---------------------------------------------------
# FLIPPING
# ---------------------------------------------------

# Horizontal flip
flipped_h = cv2.flip(roi, 1)

# Vertical flip
flipped_v = cv2.flip(roi, 0)

# ---------------------------------------------------
# DISPLAY
# ---------------------------------------------------

cv2.imshow("Original", img)

cv2.imshow("ROI", roi)

cv2.imshow("Flipped H", flipped_h)

cv2.imshow("Flipped V", flipped_v)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2

# Load image
img = cv2.imread(r"C:\Users\Zinkal Patel\Desktop\Images\HappyFish.jpg")

# ---------------------------------------------------
# COLOR SPACE CONVERSION
# ---------------------------------------------------

# Convert BGR to Grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Convert BGR to HSV
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# ---------------------------------------------------
# DISPLAY
# ---------------------------------------------------

cv2.imshow("Original BGR Image", img)

cv2.imshow("Grayscale Image", gray)

cv2.imshow("HSV Image", hsv)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
